```python
from modules.preprocessing import WeatherPreprocessor, process_and_merge
from modules.data_loader import load_traffic_data
```

```python
wp = WeatherPreprocessor()
tdf = load_traffic_data()
```

```python
raw = wp.load_raw()
print("RAW:", len(raw), raw.index.min(), raw.index.max())

sel = wp.select_columns(raw)
print("SELECTED:", len(sel))

dec = wp.decode(sel)
print("DECODED:", len(dec))

res = wp.resample(dec)
print("RESAMPLED:", len(res), res.index.min(), res.index.max())
```

```python
wdf = wp.run()
wdf.head()
```

```python
wdf.isnull().sum()
wdf.info()
wdf.nunique()
```

```python
merged_df = process_and_merge(traffic_df=tdf,weather_df=wdf)
```
